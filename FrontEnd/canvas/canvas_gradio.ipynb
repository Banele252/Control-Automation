### Creating a gradio app that for the interactive agent

```python
import gradio as gr

```

```python
def greet(name, intensity):
    return "Hello, " + name + "!" * int(intensity)

demo = gr.Interface(
    fn=greet,
    inputs=["text", "slider"],
    outputs=["text"],
    api_name="predict"
)

```

```python
def chat(message, history):
    if history is None:
        history = []
    response = "You said: " + message
    history.append((message, response))
    return history, history
```

```python
import requests


def chat(message, history):
    messages = message
    BaseURL = "http://localhost:8000"
    endpoint = "/interactive-agent/interactive-agent"
    url = BaseURL + endpoint
    params = {
        "message": messages
    }
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()
        data = response.json()
        return data.get('result', 'No result found')
    except requests.exceptions.RequestException as e:
        print(f"Error occurred: {e}")
        return "Error occurred while fetching response"

#final = chat("How many exceptions are there?")
#print(final)
```

```python
demo = gr.Interface(
    fn=chat,
    inputs=gr.Textbox(lines=2, placeholder="Enter your message here..."),
    outputs="text",
    api_name="predict"
)
```

```python
demo = gr.ChatInterface(
    fn=chat
)
```

```python
demo.launch()
```
